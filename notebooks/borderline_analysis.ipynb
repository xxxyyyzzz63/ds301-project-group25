Uncertainty (Borderline) Analysis

This notebook analyzes the model’s uncertainty by examining test samples with AI-likeness scores close to 50.

While previous stress testing revealed systematic misclassification under extreme stylistic conditions, this analysis focuses on naturally occurring ambiguous cases in the test set.

These borderline examples provide insight into the model’s decision boundary and help identify the types of inputs where the detector struggles to make confident predictions.

```python
import pickle
import pandas as pd

with open("../models/updated_week4_week5_outputs.pkl", "rb") as f:
    saved = pickle.load(f)

df = saved["test_results_df"]

borderline = df.assign(distance_to_50=(df["ai_likeness_score"] - 50).abs())
borderline = borderline.sort_values("distance_to_50").head(20)

display(borderline[[
    "review_text", "true_label", "predicted_label",
    "ai_probability", "ai_likeness_score",
    "uncertainty_band", "top_features"
]])
```

```python
for text in borderline["review_text"].head(5):
    print(text)
    print("-" * 50)
```

```python
print("Average score of borderline:", borderline["ai_likeness_score"].mean())

extreme = df[(df["ai_likeness_score"] < 10) | (df["ai_likeness_score"] > 90)]
print("Average score of extreme:", extreme["ai_likeness_score"].mean())
```

Connection to Stress Test:

While stress testing reveals systematic errors under controlled conditions, borderline analysis shows how the model behaves on naturally ambiguous data.

Together, these analyses provide a more complete understanding of the model’s limitations.
